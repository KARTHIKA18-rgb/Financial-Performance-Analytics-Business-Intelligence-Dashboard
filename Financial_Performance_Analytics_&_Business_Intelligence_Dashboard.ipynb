# **Financial Performance Analytics & Business Intelligence Dashboard**

### Project Summary



Financial Performance Analytics & Business Intelligence Dashboard is a data analytics project designed to analyze an organization's financial performance using key business metrics such as revenue, profit, return on investment (ROI), profit margin, revenue growth, and operating expenses. The project involves data cleaning, preprocessing, exploratory data analysis (EDA), and interactive dashboard development to identify financial trends, business performance, and growth opportunities. Through informative visualizations and KPI analysis, the dashboard enables users to monitor revenue and profitability, compare financial performance across different business segments, identify high-performing and underperforming areas, and support strategic decision-making. The project provides actionable business insights that help organizations improve operational efficiency, optimize financial planning, and drive sustainable business growth.


## Project Objectives

* To monitor key financial indicators (KPIs) such as Revenue, Net Profit, Profit Margin, ROI, Revenue Growth, and Operating Expenses.
* To generate actionable business insights that help improve profitability, optimize costs, and support strategic business planning.
* To develop an end-to-end Financial Performance Analytics Dashboard** that enables stakeholders to track financial health, evaluate business performance, and make data-driven decisions effectively.


### Dataset Description

The **Financial Performance** dataset contains financial and business performance metrics for different business entities across multiple industry sectors. Each record represents a business or investment entity and includes key financial indicators such as **Revenue, Profit per Unit, Return on Assets (ROA), Return on Equity (ROE), Return on Invested Capital (ROIC), Price-to-Earnings (P/E) Ratio, Price-to-Book (P/B) Ratio, Price-to-Sales (P/S) Ratio, Dividend Yield, Sales Growth, Cash Flow Growth, Book Value Growth, Operating Cost, Management Fees, and Profit Growth (Year-to-Date, 3-Year, and 5-Year)**. The dataset is designed to support financial performance analysis, profitability assessment, growth trend evaluation, and business benchmarking. It is well suited for data cleaning, exploratory data analysis (EDA), KPI monitoring, and interactive dashboard development, enabling stakeholders to identify high-performing entities, compare financial performance across sectors, evaluate operational efficiency, and generate data-driven business insights for strategic decision-making.

# ***Data Preprocessing /Feature Engineering***

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder

EntityID_encoder = LabelEncoder()
BusinessName_encoder = LabelEncoder()
BusinessSector_encoder = LabelEncoder()                                                    #  10 object into int

df['EntityID'] = EntityID_encoder.fit_transform(df['EntityID'])
df['BusinessName'] = BusinessName_encoder.fit_transform(df['BusinessName'])
df['BusinessSector'] = BusinessSector_encoder.fit_transform(df['BusinessSector'])
```

# ***Feature Relationship Analysis***

```python
plt.figure(figsize=(15, 12))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Correlation Heatmap', fontsize=16)
plt.show()
```

Correlation Heatmap Analysis

* Most variables show weak correlations, with correlation coefficients close to 0, indicating that the financial metrics are largely independent and measure different aspects of business performance.
* ROA (%) and ROIC (%) have a strong positive correlation (≈ 0.69), suggesting that companies with efficient asset utilization also tend to generate higher returns on invested capital.
* ROA (%) and ROE (%) exhibit a moderate positive correlation (≈ 0.58), indicating that improvements in asset efficiency are generally associated with higher returns to shareholders.
* ROE (%) and ROIC (%) are positively correlated (≈ 0.30), showing that companies with better equity returns often demonstrate improved investment efficiency.
* Profit Growth (YTD %) has a moderate positive correlation with P/E Ratio (≈ 0.57), implying that companies experiencing higher recent profit growth are often valued more highly by investors.
* Profit Growth (YTD %) also shows a moderate positive relationship with ROA (%) (≈ 0.45), indicating that efficient asset utilization contributes to stronger short-term profit growth.
* Dividend Yield (%) has a moderate negative correlation with Profit Growth (YTD %) (≈ -0.57), suggesting that companies paying higher dividends tend to experience comparatively lower short-term profit growth, as more earnings may be distributed rather than reinvested.
* Operating Cost has a moderate positive correlation with Management Fees (≈ 0.53), indicating that businesses with higher operational expenses also tend to incur higher management-related costs.
* Revenue has very weak correlations with most financial ratios, implying that higher revenue alone does not necessarily translate into higher profitability, efficiency, or growth.
* Business Sector and Business Name show negligible correlations with numerical financial variables, indicating that these categorical identifiers do not have a direct linear relationship with the financial performance metrics in the dataset.

**Overall The heatmap indicates that profitability and return-based metrics (ROA, ROE, ROIC, and Profit Growth) are more strongly related to one another than to revenue or categorical variables.**

*KPI* *Calculations*

```python
# 1. Total Revenue
total_revenue = df['Revenue'].sum()

# 2. Average Profit Per Unit
avg_profit_per_unit = df['ProfitPerUnit'].mean()

# 3. Average ROA
avg_roa = df['ROA_Pct'].mean()

# 4. Average ROIC
avg_roic = df['ROIC_Pct'].mean()

# 5. Average Sales Growth
avg_sales_growth = df['SalesGrowth_Pct'].mean()

# 6. Average Operating Cost
avg_operating_cost = df['OperatingCost_Pct'].mean()

# 7. Average Profit Growth (YTD)
avg_profit_growth_ytd = df['ProfitGrowth_YTD_Pct'].mean()

# ===========================
# DISPLAY KPIs
# ===========================

print("="*50)
print("      FINANCIAL PERFORMANCE KPIs")
print("="*50)

print(f" Total Revenue                 : {total_revenue:,.2f}")
print(f"Average Profit Per Unit       : {avg_profit_per_unit:.2f}")
print(f"Average ROA (%)               : {avg_roa:.2f}%")
print(f"Average ROIC (%)              : {avg_roic:.2f}%")
print(f"Average Sales Growth (%)      : {avg_sales_growth:.2f}%")
print(f"Average Operating Cost (%)    : {avg_operating_cost:.2f}%")
print(f"Average Profit Growth (YTD)   : {avg_profit_growth_ytd:.2f}%")
```
